Q1. What is a Decision Tree? Explain its structure (Root, Nodes, Leaves) with a real-life example.

A Decision Tree is a supervised machine learning algorithm used for both classification and regression tasks. It breaks down a dataset into smaller subsets while at the same time an associated decision tree is incrementally developed. The final result is a tree with decision nodes and leaf nodes.

$\text{1. Structural Components}$

- Root Node: This is the top-most node of a decision tree. It represents the entire population or dataset, which further gets divided into two or more homogeneous sets based on the most significant feature split.

- Decision Nodes (Internal Nodes): When a sub-node splits into further sub-nodes, it is called a decision node. These nodes represent tests on specific attributes or features.

- Branches / Edges: They represent the outcome of a decision or test rule, connecting one node to the next.

- Leaf Nodes (Terminal Nodes): Nodes that do not split further are called leaf or terminal nodes. They represent the final predicted class or continuous value outcome.

$\text{2. Real-Life Example}$

Imagine deciding whether to accept a job offer:
1. Root Node: Is the salary greater than $80,000?

- If No: $\rightarrow$ Leaf Node: Decline offer.
- If Yes: $\rightarrow$ Proceed to next branch.
2. Decision Node: Is the commute less than 30 minutes?

- If No: $\rightarrow$ Leaf Node: Decline offer.
- If Yes: $\rightarrow$ Proceed to next branch.
3. Decision Node: Does it offer remote flexibility?
- If Yes: $\rightarrow$ Leaf Node: Accept Offer!

Q2. Differentiate between Gini Impurity and Entropy.

Which one is used by default in Scikit-learn and why?

### **Gini Impurity vs. Entropy**

| Feature | Gini Impurity | Entropy |
| :--- | :--- | :--- |
| **Definition** | Measures the probability of a randomly chosen element being incorrectly classified if it was randomly labeled according to the distribution in the subset. | Measures the degree of randomness, chaos, or impurity in a given dataset. |
| **Mathematical Formula** | $$Gini = 1 - \sum (P_i)^2$$|$$Entropy = - \sum (P_i \cdot \log_2 P_i)$$ |
| **Range of Values** | Ranges from `0.0` (pure node) to `0.5` (equally distributed 50/50 split). | Ranges from `0.0` (pure node) to `1.0` (equally distributed 50/50 split). |
| **Computation Speed** | Faster to compute because it does not require calculating logarithmic functions. | Slower to compute due to logarithmic calculations ($log_2$). |

**Default Selection in Scikit-learn:** Scikit-learn uses **gini** as the default criterion for decision trees because it is **computationally much faster** since it avoids logarithmic calculations. This saves significant training time on large datasets while producing almost identical tree quality compared to Entropy.

Q3. What is Overfitting in Decision Trees?

How can we detect it using training and testing accuracy?

Overfitting occurs when a Decision Tree grows too deep and complex, learning the specific noise, outliers, and random fluctuations in the training data rather than the underlying general patterns. Because decision trees split data down to perfect purity if left unrestricted, they can easily memorize a dataset instead of understanding it.

$\text{Detection using Accuracy Metrics}$

We can detect overfitting by comparing the model's accuracy on the training dataset against its accuracy on a completely unseen testing dataset:

- High Training Accuracy (Often 95% to 100%): The tree has perfectly memorized the specific training examples.

- Low Testing Accuracy: The tree fails completely when exposed to new data because its rigid, hyper-specific rules do not generalize well.

- Rule of Thumb: A large gap or discrepancy between high training performance and significantly lower testing performance is a clear signal of an overfitted model.

Q4. Explain Pruning in Decision Trees.

What is the difference between Pre-pruning and Post-pruning?

Pruning is a technique used to solve overfitting. It involves cutting down or removing parts of the decision tree (branches or nodes) that contribute little to predictive power, transforming a highly complex tree into a simpler, more generalized tree.

1. Pre-pruning (Early Stopping):

- This technique halts the growth of the decision tree during the training process before it becomes overly complex.

- It utilizes hyperparameters such as max_depth (limiting how deep the tree goes), min_samples_split (minimum samples required to split a node), and min_samples_leaf.

- Pros: Highly efficient as it avoids building the complete tree structure.

2. Post-pruning (Cost Complexity Pruning):

- This technique allows the decision tree to grow completely to its maximum size and depth until all leaves are pure.

- It then works backwards, analyzing individual sub-branches and removing nodes that add minimal predictive value or increase validation error.

- Pros: More thorough because it assesses full feature relationships before removing branches, though it requires more computation.

Q5. What is Feature Importance?

How can it help businesses in decision-making?

Feature Importance is a metric calculated by the Decision Tree algorithm that scores features based on how effective they are at splitting data cleanly. A feature's importance is determined by how much total Gini impurity or Entropy it reduces across all nodes in the tree. Features that cause clean, decisive splits near the top (root) of the tree receive a higher importance score.

$\text{Business Applications & Decision-Making}$

$Resource\ Optimization:$ It tells a company exactly which variables matter. For instance, in a marketing campaign, if duration_of_call has the highest feature importance, a business knows to train their sales agents on maintaining engaging, long-lasting customer conversations.

$Cost\ Reduction:$ If certain data features cost money or time to collect but show near-zero importance, a business can safely stop collecting that data.

$Risk\ Mitigation:$ In financial lending, knowing which customer features (e.g., credit history vs. age) predict default risk allows banks to craft strict loan approval policies based exclusively on the highest impact metrics.

Q8. Data

- Convert categorical variables into numerical

- Mention which encoding method you used

Interpret:

- Why is encoding necessary for Decision Trees?

```python
import pandas as pd

# Separate independent and dependent features before encoding to avoid encoding target as a dummy column
X = df.drop(columns=['deposit'])
y = df['deposit']

# Map the target variable 'deposit' explicitly to binary values (yes=1, no=0)
y = y.map({'yes': 1, 'no': 0})

# Get all categorical columns left in X
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()
print("Categorical variables to be encoded:", categorical_cols)

# Apply One-Hot Encoding using pd.get_dummies
X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

# Display a preview of the newly encoded DataFrame features
print("\n--- Encoded Features Preview ---")
display(X_encoded.head())
print(f"\nNew shape after encoding: {X_encoded.shape}")
```

$Proper\ Interpretation$

- Encoding Method Used: We implemented One-Hot Encoding (pd.get_dummies with drop_first=True) for all multi-class categorical features and a direct manual binary mapping for our target feature (deposit).

- Why drop_first=True? It removes the first category column of each encoded feature to prevent multicollinearity (the dummy variable trap), ensuring clean, independent representations of the structural flags.

- Why is encoding necessary for Decision Trees? Scikit-learn’s implementation of the Decision Tree Classifier (DecisionTreeClassifier) operates strictly on numerical values. It evaluates mathematical inequalities (e.g., Feature <= Threshold) at every internal node split. It cannot process text descriptions directly (e.g., comparing if job <= 'management'), so converting text labels into numerical expressions is vital to establish mathematical splits.

Q9. Feature Selection & Splitting
- Separate:
  - Features (X)
  - Target (y)
- Perform train-test split (80-20)
Interpret:
- Why is train-test split important?

```python
from sklearn.model_selection import train_test_split

# Features (X_encoded) and Target (y) have already been separated in Q8.
# Let's perform the train-test split (80% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.20, random_state=42, stratify=y
)

print("--- Train-Test Split Dimensions ---")
print(f"X_train shape: {X_train.shape} (80% of independent features)")
print(f"X_test shape:  {X_test.shape} (20% of independent features)")
print(f"y_train shape: {y_train.shape} (80% of target labels)")
print(f"y_test shape:  {y_test.shape} (20% of target labels)")
```

- Why Train-Test Split is Important: Splitting the data ensures we evaluate the model's performance on completely unseen data. If we evaluate our model using the same records it used to learn rules, the metrics will be overly optimistic, rendering us unable to recognize if the tree has simply memorized patterns (overfitted). It simulates real-world deployment where the model encounters new clients.

- Why use stratify=y? Stratification ensures that the train and test subsets maintain the exact same proportion of target classes (yes vs no outcomes) as the original dataset. This prevents selection bias where one split accidentally gets significantly more positive outcomes than another, leading to stable, highly balanced training.

Q10. Model Building
- Train a Decision Tree Classifier
- Use:
  - criterion='gini'
  - max_depth=5
Interpret:
- What does max_depth control?

```python
from sklearn.tree import DecisionTreeClassifier

# Train a Decision Tree Classifier with the requested constraints
dt_classifier = DecisionTreeClassifier(criterion='gini', max_depth=5, random_state=42)
dt_classifier.fit(X_train, y_train)

print("Decision Tree Classifier successfully trained.")
print(f"Parameters used: criterion='{dt_classifier.criterion}', max_depth={dt_classifier.max_depth}")
```

$Proper\ Interpretation$

- What does max_depth control? max_depth governs the maximum height or vertical length of the decision tree. It specifies the maximum number of structural levels or splitting steps allowed from the top root node down to the deepest leaf node.

- Why do we explicitly restrict it? By setting max_depth=5, we prevent the tree from growing indefinitely and becoming overly complex. Unconstrained decision trees will continue splitting until every leaf contains entirely pure subsets, which captures unnecessary noise and leads to severe overfitting. Limiting the depth serves as a robust pre-pruning mechanism, enforcing simpler generalized boundary constraints that help the model perform reliably on unseen testing profiles.

Q11. Model Evaluation

Perform:
- Accuracy score
- Confusion Matrix
- Classification Report

Interpret (MUST):
- Is the model performing well?
- Explain Precision & Recall
- Any class imbalance issue?

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Predict on the test set
y_pred = dt_classifier.predict(X_test)

# Calculate Evaluation Metrics
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred, target_names=['No Deposit', 'Deposit'])

print(f"--- Model Evaluation Results ---")
print(f"Accuracy Score: {accuracy:.4f} ({accuracy * 100:.2f}%)")
print("\n--- Confusion Matrix ---")
print(conf_matrix)
print("\n--- Classification Report ---")
print(class_report)
```

$Proper\ Interpretation$

- Is the model performing well? Yes, an accuracy score of 79.36% indicates a robust model for a standard baseline configuration. It accurately classifies roughly 8 out of every 10 customers.

- Precision and Recall Explanation:

  - Precision (Exactness): For the Deposit (yes) class, precision is 78%. This means that when our model predicts a customer will subscribe to a deposit, it is correct 78% of the time.

  - Recall (Completeness): For the Deposit (yes) class, recall is 78%. This indicates that out of all actual customers who signed up for a deposit, the model successfully captures and identifies 78% of them.

- Class Imbalance Check: There is no class imbalance issue. Looking at the support line, the test dataset contains 1,174 "No" cases and 1,059 "Yes" cases. Because the distribution is nearly equal (~53% to ~47%), standard accuracy remains a highly trustworthy metric, and the balanced precision/recall scores confirm uniform performance across both classes.
